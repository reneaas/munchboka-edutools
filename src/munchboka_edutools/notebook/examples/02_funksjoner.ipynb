# Utforsk en andregradsfunksjon

Endre koeffisientene og undersøk hvordan grafen forandrer seg.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 200)
a, b, c = 1, -2, -3
```

```python
y = a*x**2 + b*x + c
plt.plot(x, y)
plt.axhline(0, color="gray")
plt.axvline(0, color="gray")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()
```

Finn koeffisienter som gir to, ett og ingen nullpunkter.
