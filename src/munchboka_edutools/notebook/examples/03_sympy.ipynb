# Derivasjon med SymPy

La Python finne den deriverte og sammenlign med din egen utregning.

```python
import sympy as sp

x = sp.Symbol("x")
f = x**3 - 3*x**2 + 2
f
```

```python
derivert = sp.diff(f, x)
derivert
```

```python
sp.solve(derivert, x)
```
